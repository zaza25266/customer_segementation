```python
import pandas as pd

# 1. Create the sample data
df = pd.DataFrame({
    "userId": [1, 1, 2],
    "score": [10, 20, 30]
})

print("--- Original DataFrame ---")
print(df)

# 2. Define the calculation: subtract the user's average score from their actual score
def calculate_difference(group):
    return group["score"] - group["score"].mean()

# 3. Scenario A: Default (group_keys=True)
# Notice how 'userId' is added as an extra index level.
result_default = df.groupby("userId").apply(calculate_difference)

print("\n--- Output with group_keys=True (Default) ---")
print(result_default)


# 4. Scenario B: group_keys=False
# Notice how it preserves the original flat index.
result_false = df.groupby("userId", group_keys=False).apply(calculate_difference)

print("\n--- Output with group_keys=False ---")
print(result_false)
```
